```python
!pip install ipywidgets

import ipywidgets as widgets
from IPython.display import display

# Display field
display_box = widgets.Text(
    value='',
    description='',
    disabled=False,
    layout=widgets.Layout(width='240px')
)

# Button click handler
def on_button_click(b):
    if b.description == 'C':
        display_box.value = ''
    elif b.description == '=':
        try:
            display_box.value = str(eval(display_box.value))
        except:
            display_box.value = 'Error'
    else:
        display_box.value += b.description

# Button layout
buttons = [
    '7', '8', '9', '/',
    '4', '5', '6', '*',
    '1', '2', '3', '-',
    '0', '.', 'C', '+',
    '='
]

button_widgets = []
for label in buttons:
    button = widgets.Button(
        description=label,
        layout=widgets.Layout(width='55px', height='40px')
    )
    button.on_click(on_button_click)
    button_widgets.append(button)

# Arrange buttons in a grid
rows = [
    widgets.HBox(button_widgets[i:i+4])
    for i in range(0, 16, 4)
]
rows.append(widgets.HBox([button_widgets[-1]]))

# Display calculator
display(display_box)
for row in rows:
    display(row)
```
